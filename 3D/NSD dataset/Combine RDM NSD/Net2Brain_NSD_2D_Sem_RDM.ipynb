# Net2Brain 2D and Semantic NN (NSD)

## extracting Data, RDM, and RSA

```python
# CONFIGURATION
import os

from pathlib import Path

nsd_path = Path("../Algonauts23_shared_Net2Brain")

stimuli_path = nsd_path / "images"

print("NSD dataset:", nsd_path)
print("Stimuli path:", stimuli_path)
print("Images exist:", stimuli_path.exists())
print("Number of images:", len(list(stimuli_path.glob("*.png"))))
```

```python
# Paths for RSA

nsd_path = Path("../Algonauts23_shared_Net2Brain")

# All subjects
subject_paths = sorted(nsd_path.glob("subj*"))

# Brain RDM paths
subject_rdm_paths = {
    subject.name: subject / "rdms"
    for subject in subject_paths
}

print("NSD dataset:", nsd_path)

print("\nSubjects:")
for subject in subject_paths:
    print(subject.name)

print("\nRDM paths:")
for subject, rdm_path in subject_rdm_paths.items():
    print(subject, "->", rdm_path)
```

## 2D

```python
from net2brain.feature_extraction import FeatureExtractor
from pathlib import Path

models_2d = [
    "colorization",
    "edge_texture",
    "keypoints2d",
    "segment_unsup2d"
]

for model_name in models_2d:

    print(f"\nExtracting: {model_name}")

    fx = FeatureExtractor(
        model=model_name,
        netset="Taskonomy",
        device="cuda"
    )

    save_path = f"2D_{model_name}_feats_NSD_last2"

    fx.extract(
        data_path=stimuli_path,
        save_path=save_path,
        layers_to_extract=["layer3", "layer4"]
    )

    print(f"Finished: {model_name}")
```

```python
from net2brain.rdm_creation import RDMCreator

creator = RDMCreator(
    verbose=True,
    device="cuda"
)

for model_name in models_2d:

    feature_path = f"2D_{model_name}_feats_NSD_last2"
    rdm_path = f"2D_{model_name}_RDMs_NSD_last2"

    print(f"\nCreating RDMs for: {model_name}")

    creator.create_rdms(
        feature_path=feature_path,
        save_path=rdm_path,
        save_format="npz"
    )

    print(f"Finished: {model_name}")
```

```python
import numpy as np
from pathlib import Path

models_2d_paths = {
    "Colorization": Path("2D_colorization_RDMs_NSD_last2"),
    "Edge_Texture": Path("2D_edge_texture_RDMs_NSD_last2"),
    "Segment_Unsup2D": Path("2D_segment_unsup2d_RDMs_NSD_last2"),
    "KeyPoints2D": Path("2D_keypoints2d_RDMs_NSD_last2")
}

layers = [
    "RDM_layer3.npz",
    "RDM_layer4.npz"
]

rdms = []

stimuli_names =[]


for model_name, model_path in models_2d_paths.items():

    for layer_file in layers:

        rdm_path = model_path / layer_file

        print("Loading:", rdm_path)

        data = np.load(rdm_path, allow_pickle=True)

        rdm = np.asarray(data["rdm"], dtype=float)

        rdms.append(rdm)

        if "stimuli_name" in data.files:
            stimuli_names.append(data["stimuli_name"])


# Check shapes
shapes = [rdm.shape for rdm in rdms]

print("\nRDM shapes:")
print(shapes)

if len(set(shapes)) != 1:
    raise ValueError("RDM shapes do not match!")


# Check stimulus ordering
if len(stimuli_names) > 1:

    for i in range(1, len(stimuli_names)):

        if not np.array_equal(
            stimuli_names[0],
            stimuli_names[i]
        ):
            raise ValueError(
                "Stimulus ordering differs between RDMs!"
            )


# Stack and average
rdms_stack = np.stack(rdms, axis=0)

print("\nStacked shape:", rdms_stack.shape)

rdm_2d = np.mean(rdms_stack, axis=0)

print("Combined 2D RDM shape:", rdm_2d.shape)


# Save
output_file = Path("2D_RDM_NSD.npz")

np.savez(
    output_file,
    rdm=rdm_2d,
    stimuli_name=stimuli_names[0],
    models=np.array(list(models_2d_paths.keys())),
    layers=np.array(["layer3", "layer4"])
)

print("\nSaved:", output_file)
```

## Semantic

```python
from net2brain.feature_extraction import FeatureExtractor

models_semantic = [
    "class_object",
    "class_scene",
    "segment_semantic"
]

for model_name in models_semantic:

    print(f"\nExtracting: {model_name}")

    fx = FeatureExtractor(
        model=model_name,
        netset="Taskonomy",
        device="cuda"
    )

    save_path = f"Semantic_{model_name}_feats_NSD_last2"

    fx.extract(
        data_path=stimuli_path,
        save_path=save_path,
        layers_to_extract=["layer3", "layer4"]
    )

    print(f"Finished: {model_name}")
```

```python
from net2brain.rdm_creation import RDMCreator

creator_semantic = RDMCreator(
    verbose=True,
    device="cuda"
)

for model_name in models_semantic:

    feature_path = f"Semantic_{model_name}_feats_NSD_last2"
    rdm_path = f"Semantic_{model_name}_RDMs_NSD_last2"

    print(f"\nCreating RDMs for: {model_name}")

    creator_semantic.create_rdms(
        feature_path=feature_path,
        save_path=rdm_path,
        save_format="npz"
    )

    print(f"Finished: {model_name}")
```

```python
import numpy as np
from pathlib import Path

models_semantic = {
    "class_object": Path("Semantic_class_object_RDMs_NSD_last2"),
    "class_scene": Path("Semantic_class_scene_RDMs_NSD_last2"),
    "segment_semantic": Path("Semantic_segment_semantic_RDMs_NSD_last2")
}

layers = [
    "RDM_layer3.npz",
    "RDM_layer4.npz"
]

rdms = []
stimuli_names = []

for model_name, model_path in models_semantic.items():

    for layer_file in layers:

        rdm_path = model_path / layer_file

        print("Loading:", rdm_path)

        data = np.load(rdm_path, allow_pickle=True)

        rdm = np.asarray(data["rdm"], dtype=float)

        rdms.append(rdm)

        if "stimuli_name" in data.files:
            stimuli_names.append(data["stimuli_name"])



# Check shapes
shapes = [
    rdm.shape
    for rdm in rdms
]

if len(set(shapes)) != 1:
    raise ValueError(
        f"RDM shapes differ: {shapes}"
    )


# Check stimulus ordering
if len(stimuli_names) > 1:

    for i in range(1, len(stimuli_names)):

        if not np.array_equal(
            stimuli_names[0],
            stimuli_names[i]
        ):
            raise ValueError(
                "Stimulus ordering differs between RDMs!"
            )


# Stack and average
rdms_stack = np.stack(rdms, axis=0)

print("\nStacked shape:", rdms_stack.shape)

rdm_sem = np.mean(rdms_stack, axis=0)

print("Combined Semantic RDM shape:", rdm_sem.shape)


# Save
output_file = Path("Semantic_RDM_NSD.npz")

np.savez(
    output_file,
    rdm=rdm_sem,
    stimuli_name=stimuli_names[0],
    models=np.array(list(models_semantic.keys())),
    layers=np.array(["layer3", "layer4"])
)

print("\nSaved:", output_file)
```
