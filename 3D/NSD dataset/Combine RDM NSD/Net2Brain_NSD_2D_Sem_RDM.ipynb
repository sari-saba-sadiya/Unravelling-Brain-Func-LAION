## extracting Data, RDM, and RSA

```python
# CONFIGURATION
import os

from pathlib import Path

nsd_path = Path("../Algonauts23_shared_Net2Brain")

stimuli_path = nsd_path / "images"

print("NSD dataset:", nsd_path)
print("Stimuli path:", stimuli_path)
print("Images exist:", stimuli_path.exists())
print("Number of images:", len(list(stimuli_path.glob("*.png"))))
```

## 2D

```python
from net2brain.feature_extraction import FeatureExtractor

models_2d = [
    "colorization",
    "denoising",
    "edge_texture",
    "inpainting",
    "keypoints2d",
    "segment_unsup2d"
]

layers_2d = [
    "relu1",
    "layer4"
]

for model_name in models_2d:

    print(f"\nExtracting: {model_name}")

    fx = FeatureExtractor(
        model=model_name,
        netset="Taskonomy",
        device="cuda"
    )

    save_path = (
        f"2D_{model_name}_feats_NSD_relu1_layer4"
    )

    fx.extract(
        data_path=stimuli_path,
        save_path=save_path,
        layers_to_extract=layers_2d
    )

    print(f"Finished: {model_name}")
```

```python
from net2brain.rdm_creation import RDMCreator

creator = RDMCreator(
    verbose=True,
    device="cuda"
)

for model_name in models_2d:

    feature_path = (
        f"2D_{model_name}_feats_NSD_relu1_layer4"
    )

    rdm_path = (
        f"2D_{model_name}_RDMs_NSD_relu1_layer4"
    )

    print(
        f"\nCreating RDMs for: {model_name}"
    )

    creator.create_rdms(
        feature_path=feature_path,
        save_path=rdm_path,
        save_format="npz"
    )

    print(f"Finished: {model_name}")
```

```python
import numpy as np
from pathlib import Path

# 2D TASKONOMY RDM COMBINATION
models_2d_paths = {
    "Colorization":
        Path("2D_colorization_RDMs_NSD_relu1_layer4"),

    "Denoising":
        Path("2D_denoising_RDMs_NSD_relu1_layer4"),

    "Edge_Texture":
        Path("2D_edge_texture_RDMs_NSD_relu1_layer4"),

    "Inpainting":
        Path("2D_inpainting_RDMs_NSD_relu1_layer4"),

    "KeyPoints2D":
        Path("2D_keypoints2d_RDMs_NSD_relu1_layer4"),

    "Segment_Unsup2D":
        Path("2D_segment_unsup2d_RDMs_NSD_relu1_layer4")
}

layers = [
    "RDM_relu1.npz",
    "RDM_layer4.npz"
]

rdms = []
stimuli_reference = None

# LOAD ALL RDMs
for model_name, model_path in models_2d_paths.items():

    for layer_file in layers:

        rdm_path = model_path / layer_file

        print("Loading:", rdm_path)

        if not rdm_path.exists():
            raise FileNotFoundError(
                f"File not found: {rdm_path.resolve()}"
            )

        data = np.load(
            rdm_path,
            allow_pickle=True
        )

        rdm = np.asarray(
            data["rdm"],
            dtype=float
        )

        rdms.append(rdm)

        # Check stimulus ordering
        if "stimuli_name" in data.files:

            stimuli = data["stimuli_name"]

            if stimuli_reference is None:

                stimuli_reference = stimuli

            elif not np.array_equal(
                stimuli_reference,
                stimuli
            ):

                raise ValueError(
                    f"Stimulus ordering differs in "
                    f"{rdm_path}"
                )

# CHECK SHAPES
shapes = [
    rdm.shape
    for rdm in rdms
]

print("\nRDM shapes:")

for i, shape in enumerate(shapes, 1):
    print(f"RDM {i}: {shape}")

if len(set(shapes)) != 1:
    raise ValueError(
        f"RDM shapes do not match: {shapes}"
    )

# Z-SCORE EACH RDM

def zscore_rdm(rdm):

    std = np.std(rdm)

    if std == 0:
        raise ValueError(
            "RDM has zero standard deviation."
        )

    return (
        (rdm - np.mean(rdm))
        / std
    )


rdms = [
    zscore_rdm(rdm)
    for rdm in rdms
]

# STACK AND AVERAGE
rdms_stack = np.stack(
    rdms,
    axis=0
)

print(
    "\nStacked shape:",
    rdms_stack.shape
)

rdm_2d = np.mean(
    rdms_stack,
    axis=0
)

print(
    "Combined 2D RDM shape:",
    rdm_2d.shape
)

# SAVE
output_file = Path(
    "2D_RDM_NSD.npz"
)

save_kwargs = {
    "rdm": rdm_2d,
    "models": np.array(
        list(models_2d_paths.keys())
    ),
    "layers": np.array(
        ["relu1", "layer4"]
    )
}

if stimuli_reference is not None:
    save_kwargs["stimuli_name"] = stimuli_reference

np.savez(
    output_file,
    **save_kwargs
)

print(
    "\nSaved:",
    output_file.resolve()
)

print("\nVerification:")
print("Min:", np.min(rdm_2d))
print("Max:", np.max(rdm_2d))
print("Mean:", np.mean(rdm_2d))
print("Std:", np.std(rdm_2d))
print("NaN:", np.isnan(rdm_2d).any())
print("Inf:", np.isinf(rdm_2d).any())
```

## Semantic

```python
from net2brain.feature_extraction import FeatureExtractor

models_semantic = [
    "class_object",
    "class_scene",
    "segment_semantic"
]

layers_semantic = [
    "relu1",
    "layer4"
]

for model_name in models_semantic:

    print(f"\nExtracting: {model_name}")

    fx = FeatureExtractor(
        model=model_name,
        netset="Taskonomy",
        device="cuda"
    )

    save_path = (
        f"Semantic_{model_name}_feats_NSD_relu1_layer4"
    )

    fx.extract(
        data_path=stimuli_path,
        save_path=save_path,
        layers_to_extract=layers_semantic
    )

    print(f"Finished: {model_name}")
```

```python
from net2brain.rdm_creation import RDMCreator

creator_semantic = RDMCreator(
    verbose=True,
    device="cuda"
)

for model_name in models_semantic:

    feature_path = (
        f"Semantic_{model_name}_feats_NSD_relu1_layer4"
    )

    rdm_path = (
        f"Semantic_{model_name}_RDMs_NSD_relu1_layer4"
    )

    print(
        f"\nCreating RDMs for: {model_name}"
    )

    creator_semantic.create_rdms(
        feature_path=feature_path,
        save_path=rdm_path,
        save_format="npz"
    )

    print(f"Finished: {model_name}")
```
